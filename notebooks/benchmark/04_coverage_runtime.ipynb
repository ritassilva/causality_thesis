```python
from pathlib import Path
import sys, os, json
ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "experiments/run_model.py").is_file()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside the repository.")
os.environ.setdefault("MPLCONFIGDIR", str(ROOT / ".cache/matplotlib"))
sys.path[:0] = [str(ROOT / "analysis"), str(ROOT / "experiments")]
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
import final_visualization as fv
import result_tables as th
fv.apply_style()
sources = th.load_aligned_sources()
main = sources["benchmark_with_biological"]
complete = sources["complete"]
```

## Completed and failed fits


```python
runtime_tables = {}
# Coverage and runtime expose failed fits separately from successful timing.
for name, frame in [("benchmark", main), ("tuned_untuned", complete)]:
    keys = ["dataset", "model"] + (["arm"] if name == "tuned_untuned" else [])
    coverage = (
        frame.groupby(keys)
        .agg(recorded=("finished", "size"), completed=("finished", "sum"))
        .reset_index()
    )
    coverage["completion_fraction"] = coverage.completed / coverage.recorded
    display(coverage.head(20))
    timings = (
        frame.loc[frame.finished]
        .groupby(keys + ["dataset_seed"])
        .runtime_seconds.mean()
        .reset_index()
    )
    runtime_tables[name] = timings.groupby(keys).runtime_seconds.agg(["mean", "std", "size"]).reset_index()
```

```python
for study in ("benchmark", "tuned_untuned"):
    print(study, "runtime among completed fits")
    display(runtime_tables[study])
```
